```python
# ------------------------------------------------------------
# STEP 2: Import Libraries
# ------------------------------------------------------------

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

```python
# ------------------------------------------------------------
# STEP 3: Load the Kaggle Dataset
# ------------------------------------------------------------


file_path = "/content/drive/MyDrive/Datasets/Fraud.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
```

Dataset: Fraud Detection
---

Kaggle link: https://www.kaggle.com/datasets/dhanushnarayananr/credit-card-fraud/data

---
Description: Takes previous data from card history of customers to predict if fraud is occurring

---

Target Prediction: Fraud / Not Fraud (1,000,000 rows, 8 columns)

```python
# ------------------------------------------------------------
# STEP 4: Look at the Dataset
# ------------------------------------------------------------

print(df.head())
```

Each row represents a new customer entry. The most important columns/features are the two distance features and the ratio to median purchase price. They are likely to be the most helpful in predicting fraud because a fluctuation in price or distance from usual purchasing locations is a good indicator that the card was stolen. The target variable is the "fraud" variable. There are approximately 1,000,000 entries total with no missing variables. All the columns are relatively useful in this prediction, mostly because there are so many entries that the more data available will produce better results. With smaller sets, the extra, not as important data might interfere, but it should help in this case.

```python
print(df.columns)
```

```python
# ------------------------------------------------------------
# STEP 5: Understand the Dataset
# ------------------------------------------------------------

print(df.info())
```

```python
print(df.describe())
```

```python
# ------------------------------------------------------------
# STEP 6: Select Features and Target
# ------------------------------------------------------------

features = [
    "distance_from_home",
    "distance_from_last_transaction",
    "ratio_to_median_purchase_price",
    "repeat_retailer",
    "used_chip",
    "used_pin_number",
    "online_order"
]


X = df[features]

y = df["fraud"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

Features: "distance_from_home", "distance_from_last_transaction" ,"ratio_to_median_purchase_price", "repeat_retailer", "used_chip",
"used_pin_number", "online_order". The combination of all these features monitors what a customers normal activity looks like on their account and uses that to identify abnormalities as fraud being likely to occur.
Target: The features produce Fraud (1) / Not Fraud (0) as an output.

```python
# ------------------------------------------------------------
# STEP 7: Check for Missing Values
# ------------------------------------------------------------

print(X.isnull().sum())
```

No missing values are present, so no code is needed to handle any missing values. There is no need to address any unusual variables as all data is numeric and calculable by the AI. As for any duplicate records, there may be customer accounts that look virtually identical, especially with so many data points but that will only add to the AI's predictive capability. The customers may be doing similar things, but each entry should be unique. There are only 7 columns in a dataset of 1,000,000, so there are most likely no columns that are being wasted in the AI prediction. If anything, it is a low amount of columns for the amount of data. These should address any data quality issues present here.

```python
# ------------------------------------------------------------
# STEP 9: Split Data into Training and Testing Sets
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", len(X_train))
print("Testing data:", len(X_test))
```

The data is divided into 80% (800,000) training and 20% (200,000) testing. The testing data must be separate and different from the training data so the model can learn to actually predict values when presented with data it has not seen before. It must not simply memorize. but instead recognize patterns.

```python
# ------------------------------------------------------------
# STEP 10: Create Logistic Regression Model
# ------------------------------------------------------------

model = LogisticRegression(max_iter=1000)

print(model)
```

```python
# ------------------------------------------------------------
# STEP 11: Train the Model
# ------------------------------------------------------------

model.fit(X_train, y_train)

print("Model training completed!")
```

fit() is where the machine learning happens.
 The model looks at
 Features + Known Answers
        ↓
   Learn patterns
        ↓
   Logistic Model

```python
# ------------------------------------------------------------
# STEP 12: Make Predictions
# ------------------------------------------------------------

y_pred = model.predict(X_test)

print("Predictions:")
print(y_pred[:20])
```

The model shows a few predictions where 0 = Not Fraud and 1 = Fraud. The logistic regression model is taking all the features and compiling them into trends to predict a binary result. It is ideal for this situation because There is a large amount of numeric data that can be fed into the model and we want the model to predict the binary result of Fraud/Not Fraud which exactly fits what logistic regression is made for.

```python
# ------------------------------------------------------------
# STEP 13: Calculate Accuracy
# ------------------------------------------------------------

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)

print("Accuracy percentage:", accuracy * 100, "%")
```

The model has a high accuracy of ~ 96%

```python
# ------------------------------------------------------------
# STEP 14: Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

[TP, FP, FN, TN] -> [181280, 1277, 6991, 10452]. Proportionally, the confusion matrix has a high amount of True Positives which is exactly what we want, however the False Negatives are a lot higher compared to the False Positives. In the field of fraud detection it would be much better to have more False Positives than Negatives since without that, a lot of fraud is slipping through without being detected.

```python
# ------------------------------------------------------------
# STEP 15: Classification Report
# ------------------------------------------------------------

print(classification_report(y_test, y_pred))
```

The precision metrics (~90%) for the model are fairly good, however precision is not the most important thing needed for this dataset. The recall percent is way to low for this model to be effectively doing what is needed. 60% recall means that a lot of fraud is happening without the model realizing. The f-1 score (72%) reflects a little better, but looking at individual recall and precision just shows that precision is simply bringing that average up to make the f-1 score look better, when in reality the combined score does not matter that much.

```python
# ------------------------------------------------------------
# STEP 16: Predict Probabilities
# ------------------------------------------------------------

probabilities = model.predict_proba(X_test)

print(probabilities[:10])
```

```python
# ------------------------------------------------------------
# STEP 17: Predict a New Customer
# ------------------------------------------------------------

new_customer = pd.DataFrame({
     "distance_from_home": [10],
    "distance_from_last_transaction": [20],
    "ratio_to_median_purchase_price": [10],
    "repeat_retailer": [1],
    "used_chip": [0],
    "used_pin_number": [0],
    "online_order": [0]

})

prediction = model.predict(new_customer)

probability = model.predict_proba(new_customer)

print("Prediction:", prediction)

print("Probability of No Fraud:",
      probability[0][0])

print("Probability of Fraud:",
      probability[0][1])
```

Based on the entered features (relatively unusual distance and money spent) the model predicts 85% of no fraud and 15% of fraud. Because the features are only moderately unusual the model still places 15% chance of fraud occurring but still determines no fraud based on the 85% chance.

```python
# ------------------------------------------------------------
# STEP 18: Human-Friendly Prediction
# ------------------------------------------------------------

if prediction[0] == 1:
    print("The model predicts: Fraud")
else:
    print("The model predicts: No Fraud")
```

```python
# ------------------------------------------------------------
# STEP 19: Examine Model Coefficients
# ------------------------------------------------------------

coefficients = pd.DataFrame({
    "Feature": features,
    "Coefficient": model.coef_[0]
})

print(coefficients)
```

For the model predicting fraud/not fraud, the performance was fairly well, but still significantly brought down by the low recall. The main aim in this scenario was to get that recall as high as possible, but it ended up the lowest statistic. The confusion matrix showed a good percentage of True Positives which was good for the model, but it still did not change the fact that recall was low. To be truly ready to be used in the real world, the confusion matrix had too many inconsistencies with a lot of False Negatives present.

Limitation 1: Similar to the prediction I created that resulted in 85% not fraud and 15% fraud, the model leaves too much room for error. Likely in the real world there would be some sort of percentage threshold that activates a flag on an account that fraud *might* be occuring on. This is why there are too many False Negatives

Limitation 2: The model may be interpreting the variables together and seeing trends where there are none. The variables like "used chip" and "used pin number" should be weighted at a lot less importantance than things like "average transaction price" and that would skew the models predictions.

What to change: For a dataset of 1,000,000 entries, 7 features may not be enough to fully use all of the data to the best of the models ability. It would benefit from new features being added, hopefully up to the point where recall reaches a higher point than it is currently at.
